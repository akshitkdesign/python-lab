```python
class Student:
    school_name = "UPES"

    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Aman", 20)
student2 = Student("Riya", 21)

print("School:", student1.school_name)
print("School:", student2.school_name)

print("Student 1:", student1.name, student1.age)
print("Student 2:", student2.name,student2.age)
```

```python
class Employee:
    company = "ABC Ltd"

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def show_details(self):
        print("Name:", self.name, "| Salary:", self.salary, "| Company:", Employee.company)

    @classmethod
    def change_company(cls, new_name):
        cls.company = new_name


emp1 = Employee("Aman", 50000)
emp2 = Employee("Riya", 60000)

print("Before changing company:")
emp1.show_details()
emp2.show_details()

Employee.change_company("XYZ Ltd")

print("\nAfter changing company:")
emp1.show_details()
emp2.show_details()
```

```python
class MathHelper: 

    @staticmethod
    def is_even(num):
        return num % 2 == 0
    
print(MathHelper.is_even(8))
print(MathHelper.is_even(11))
```

```python
class BankAccount:
    
    def __init__(self, name, balance):
        self.name = name
        
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance")

    def show_balance(self):
        print("Current Balance:", self.__balance)


acc = BankAccount("Aman", 1000)

acc.deposit(500)
acc.show_balance()

acc.withdraw(300)
acc.show_balance()

acc.withdraw(1500)   
```
